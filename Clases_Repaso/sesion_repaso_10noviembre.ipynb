# Sistema de gestión de un parque de atracciones

```python
# Apartado 1: Crear la clase Atraccion:
class Atraccion:

    def __init__(self, codigo, nombre, categoria, capacidad):
        self.codigo = codigo
        self.nombre = nombre
        self.categoria = categoria
        self.capacidad = capacidad
        self.ocupantes = 0
        self.abierta = True

    def mostrar_info(self):

        if self.abierta:
            estado = "Abierta"
        else:
            estado = "Cerrada"

        print(f"{self.codigo} - {self.nombre}")
        print(f"Categoria: {self.categoria}")
        print(f"Estado: {estado}")
        print(f"Ocupacion: {self.ocupantes}/{self.capacidad}")
        print("-----------------------------")


    def subir_visitantes(self, cantidad):

        if not self.abierta:
            return False

        if cantidad <= 0:
            return False

        if self.ocupantes + cantidad > self.capacidad:
            return False

        self.ocupantes += cantidad
        return True

    
    def bajar_visitantes(self, cantidad):

        if cantidad <= 0:
            return False

        if cantidad > self.ocupantes:
            return False

        self.ocupantes -= cantidad
        return True        
```

```python
# Apartado 2: Crear la clase Parque:

class Parque:

    def __init__(self):
        self.atracciones = []


    def buscar_atraccion(self, codigo):

        for atraccion in self.atracciones:
            if atraccion.codigo == codigo:
                return atraccion

        return None


    def agregar_atraccion(self, atraccion):
        
        if self.buscar_atraccion(atraccion.codigo) is not None:
            return False

        self.atracciones.append(atraccion)
        return True


    def eliminar_atraccion(self, codigo):

        atraccion = self.buscar_atraccion(codigo)

        if atraccion is None:
            return False

        if atraccion.ocupantes > 0:
            return False

        self.atracciones.remove(atraccion)
        return True

    
    def mostrar_atracciones(self):

        if len(self.atracciones) == 0:
            print("No hay atracciones registradas en el parque.")
        else:
            for atraccion in self.atracciones:
                atraccion.mostrar_info()


# Apartado 3: Ampliar la clase Parque:

    def entrar_atraccion(self, codigo, cantidad):

        atraccion = self.buscar_atraccion(codigo)

        if atraccion is None:
            return False

        return atraccion.subir_visitantes(cantidad)

    
    def salir_atraccion(self, codigo, cantidad):

        atraccion = self.buscar_atraccion(codigo)

        if atraccion is None:
            return False

        return atraccion.bajar_visitantes(cantidad)


    def cerrar_atraccion(self, codigo):

        atraccion = self.buscar_atraccion(codigo)

        if atraccion is None:
            return False

        if not atraccion.abierta:
            return False

        if atraccion.ocupantes > 0:
            return False

        atraccion.abierta = False
        return True


    def abrir_atraccion(self, codigo):

        atraccion = self.buscar_atraccion(codigo)

        if atraccion is None:
            return False

        if atraccion.abierta:
            return False

        atraccion.abierta = True
        return True

# Apartado 4: Ampliar la clase Parque (Busquedas, filtros y estadisticas):

    def obtener_disponibles(self):

        l_disponibles = []

        for atraccion in self.atracciones:
            if atraccion.abierta and atraccion.ocupantes < atraccion.capacidad:
                l_disponibles.append(atraccion)

        return l_disponibles


    def buscar_por_categoria(self, texto):

        l_resultados = [] 

        for atraccion in self.atracciones:
            if texto.lower() in atraccion.categoria.lower():
                l_resultados.append(atraccion)

        return l_resultados


    def atraccion_mas_ocupada(self):

        if len(self.atracciones) == 0:
            return None

        mayor = self.atracciones[0]

        for atraccion in self.atracciones:
            if atraccion.ocupantes > mayor.ocupantes:
                mayor = atraccion


    def mostrar_resumen(self):

        total = len(self.atracciones)
        abiertas = 0
        visitantes = 0
        plazas_libres = 0

        for atraccion in self.atracciones:
            visitantes += atraccion.ocupantes
            
            if atraccion.abierta: 
                abiertas += 1
                plazas_libres += (atraccion.capacidad - atraccion.ocupantes)

        print(f"Total de atracciones del parque: {total}")
        print(f"Atracciones abiertas: {abiertas}")
        print(f"Atracciones cerradas: {total - abiertas}")
        print(f"Visitantes en atracciones: {visitantes}")
        print(f"Plazas libres: {plazas_libres}")
```

```python
parque = Parque()

atraccion1 = Atraccion("A001", "Dragón Rojo", "Montaña rusa", 24)
atraccion2 = Atraccion("A002", "Río Salvaje", "Acuática", 16)
atraccion3 = Atraccion("A003", "Carrusel Mágico", "Infantil", 30)
atraccion4 = Atraccion("A004", "Torre del Terror", "Caída libre", 12)
atraccion5 = Atraccion("A005", "Montaña Azul", "Montaña rusa", 20)

print("REGISTRO DE ATRACCIONES")

parque.agregar_atraccion(atraccion1)
parque.agregar_atraccion(atraccion2)
parque.agregar_atraccion(atraccion3)
parque.agregar_atraccion(atraccion4)
parque.agregar_atraccion(atraccion5)

repetida = Atraccion("A001", "Otra atracción", "Infantil", 10)
print("Añadir repetida:", parque.agregar_atraccion(repetida))

print("\nCATÁLOGO INICIAL")
parque.mostrar_atracciones()

print("\nENTRADA DE VISITANTES")
print("Dragón Rojo:", parque.entrar_atraccion("A001", 15))
print("Río Salvaje:", parque.entrar_atraccion("A002", 10))
print("Carrusel Mágico:", parque.entrar_atraccion("A003", 25))
print("Torre del Terror:", parque.entrar_atraccion("A004", 8))

print("\nPRUEBAS DE CAPACIDAD")
print("Entrar 20 más en A001:", parque.entrar_atraccion("A001", 20))
print("Entrar 9 más en A001:", parque.entrar_atraccion("A001", 9))
print("Entrar 1 más en A001:", parque.entrar_atraccion("A001", 1))

print("\nSALIDA DE VISITANTES")
print("Sacar 15 de A002:", parque.salir_atraccion("A002", 15))
print("Sacar 10 de A002:", parque.salir_atraccion("A002", 10))

print("\nATRACCIONES DISPONIBLES")
disponibles = parque.obtener_disponibles()

for atraccion in disponibles:
    atraccion.mostrar_info()

print("\nBÚSQUEDA POR CATEGORÍA")
resultados = parque.buscar_por_categoria("montaña")

for atraccion in resultados:
    atraccion.mostrar_info()

print("\nCIERRE DE ATRACCIONES")
print("Cerrar A002 ocupada:", parque.cerrar_atraccion("A002"))
print("Vaciar A002:", parque.salir_atraccion("A002", 0))
print("Sacar 10 de A002:", parque.salir_atraccion("A002", 10))
print("Cerrar A002:", parque.cerrar_atraccion("A002"))
print("Entrar con A002 cerrada:", parque.entrar_atraccion("A002", 5))

print("\nREAPERTURA")
print("Abrir A002:", parque.abrir_atraccion("A002"))
print("Entrar 5 en A002:", parque.entrar_atraccion("A002", 5))

print("\nATRACCIÓN MÁS OCUPADA")
mayor = parque.atraccion_mas_ocupada()

if mayor is not None:
    mayor.mostrar_info()

print("\nELIMINACIÓN DE ATRACCIONES")
print("Eliminar A005 vacía:", parque.eliminar_atraccion("A005"))
print("Eliminar A001 ocupada:", parque.eliminar_atraccion("A001"))

print("\nESTADO FINAL DEL PARQUE")
parque.mostrar_atracciones()

print("\nESTADÍSTICAS FINALES")
parque.mostrar_resumen()
```
