## Idea general: “transformar” y “filtrar”

En muchos problemas de datos, haces dos cosas una y otra vez:

1) **Transformar**: convertir una lista en otra (dobles, cuadrados, normalización de texto, etc.).  
2) **Filtrar**: quedarte con los elementos que cumplen una condición.

La versión “clásica” usa bucles y `append`.  
La versión “funcional” suele usar:
- **comprehensions**
- `map` (transformar)
- `filter` (filtrar)
- `lambda` (función pequeña inline)

En Python, no se trata de “usar funcional siempre”, sino de elegir la versión más clara.

## List comprehensions

Una **list comprehension** es, normalmente, la alternativa más legible a `map`/`filter` cuando quieres una lista.

### 1) Transformar (dobles)

**Versión imperativa (bucle):**

```python
lista = [10, 20, 30, 40, 50]
dobles = []
for num in lista:
    dobles.append(num * 2)

print("lista:", lista)
print("dobles:", dobles)
```

**Versión con list comprehension:**

```python
lista = [10, 20, 30, 40, 50]
dobles = [num * 2 for num in lista]

print("lista:", lista)
print("dobles:", dobles)
```

### 2) Filtrar (mayores que 30)

Fíjate en el patrón: “si cumple, lo guardo”.

**Versión imperativa (bucle):**

```python
lista = [10, 20, 30, 40, 50]
lista_filtrada = []
for i in lista:
    if i > 30:
        lista_filtrada.append(i)

print("lista:", lista)
print("lista_filtrada:", lista_filtrada)
```

**Versión con list comprehension:**

```python
lista = [10, 20, 30, 40, 50]
lista_filtrada = [i for i in lista if i > 30]

print("lista:", lista)
print("lista_filtrada:", lista_filtrada)
```

### 3) Transformar con condición (notas → texto)

Caso realista: convertir notas numéricas en etiquetas.

**Versión imperativa (bucle):**

```python
notas = [2, 5, 10, 8, 6]
notas_txt = []
for n in notas:
    if n >= 5:
        notas_txt.append("aprobado")
    else:
        notas_txt.append("suspenso")

print("notas:", notas)
print("notas_txt:", notas_txt)
```

**Versión con list comprehension (operador ternario):**

```python
notas = [2, 5, 10, 8, 6]
notas_txt = ["aprobado" if n >= 5 else "suspenso" for n in notas]

print("notas:", notas)
print("notas_txt:", notas_txt)
```

## `lambda`

Una `lambda` es una **función anónima** (sin nombre) y normalmente **muy corta**.

Útil para pasar una función como argumento (`map`, `filter`, `sorted`, etc.).
Si la lógica crece, mejor define una función normal con `def`.

### 1) Función normal vs lambda

```python
def sumar(x, y):
    return x + y

print(sumar(2, 5))
```

```python
sumar = lambda x, y: x + y
print(sumar(2, 5))
```

También puedes invocar una lambda directamente (no es lo más habitual):

```python
print((lambda x, y: x + y)(2, 5))
```

### 2) Ejemplo de normalización de texto

Caso realista: normalizar nombre y apellido.

```python
nombre_completo = lambda nombre, apellido: f"Nombre completo {nombre.title()} {apellido.title()}"
print(nombre_completo("cristian", "RODRIGUEZ"))
```

## `filter`

`filter(func, iterable)` devuelve un **iterador** con los elementos para los que `func(elemento)` es `True`.

En la práctica, casi siempre harás `list(filter(...))` para materializar el resultado.

### Filtrar mayores que 4

**Versión imperativa (bucle):**

```python
lista_numeros = [2, 5, 4, 7, 9, 10, 3, 1, 6, 8]

lista_filtrada = []
for num in lista_numeros:
    if num > 4:
        lista_filtrada.append(num)

print(lista_filtrada)
```

**Versión con `filter` + lambda:**

```python
lista_numeros = [2, 5, 4, 7, 9, 10, 3, 1, 6, 8]

lista_filtrada_iter = filter(lambda num: num > 4, lista_numeros)
print(lista_filtrada_iter)          # es un iterador
print(list(lista_filtrada_iter))    # lo convertimos a lista
```

**Versión más compacta:**

```python
lista_numeros = [2, 5, 4, 7, 9, 10, 3, 1, 6, 8]
lista_filtrada = list(filter(lambda num: num > 4, lista_numeros))
print(lista_filtrada)
```

También puedes usar una función normal en vez de lambda (a veces es más legible).

```python
def mayor_que_4(num):
    return num > 4

lista_numeros = [2, 5, 4, 7, 9, 10, 3, 1, 6, 8]
print(list(filter(mayor_que_4, lista_numeros)))
```

## `map`

`map(func, iterable)` devuelve un **iterador** con el resultado de aplicar `func` a cada elemento.

### Elevar al cuadrado

**Versión imperativa (bucle):**

```python
lista_numeros = [2, 5, 4, 7, 9, 10, 3, 1, 6, 8]

lista_resultado = []
for num in lista_numeros:
    lista_resultado.append(num * num)

print(lista_resultado)
```

**Versión con `map` + lambda:**

```python
lista_numeros = [2, 5, 4, 7, 9, 10, 3, 1, 6, 8]

lista_resultado_iter = map(lambda num: num * num, lista_numeros)
print(lista_resultado_iter)        # iterador
print(list(lista_resultado_iter))  # lista
```

**Versión más compacta:**

```python
lista_numeros = [2, 5, 4, 7, 9, 10, 3, 1, 6, 8]
lista_resultado = list(map(lambda num: num * num, lista_numeros))
print(lista_resultado)
```

También puedes usar una función normal.

```python
def cuadrado(num):
    return num * num

lista_numeros = [2, 5, 4, 7, 9, 10, 3, 1, 6, 8]
print(list(map(cuadrado, lista_numeros)))
```

## Combinando `map` + `filter`

Patrón típico:
1) filtrar
2) transformar

Ejemplo: quedarnos con los números > 4 y luego elevarlos al cuadrado.

**Versión imperativa (bucle):**

```python
lista_numeros = [2, 5, 4, 7, 9, 10, 3, 1, 6, 8]

resultado = []
for num in lista_numeros:
    if num > 4:
        resultado.append(num * num)

print(resultado)
```

**Versión funcional (`filter` + `map`):**

```python
lista_numeros = [2, 5, 4, 7, 9, 10, 3, 1, 6, 8]

resultado = list(map(lambda num: num * num,
                     filter(lambda num: num > 4, lista_numeros)))

print(resultado)
```

## Comprehensions vs `map/filter`

En Python, muchas veces las comprehensions son más legibles:

- `map + filter`: útil para *componer* transformaciones y cuando ya estás en estilo funcional.
- `list comprehension`: suele ser la opción más clara para “de una lista a otra”.

Mismo ejemplo anterior con comprehension:

```python
lista_numeros = [2, 5, 4, 7, 9, 10, 3, 1, 6, 8]
resultado = [num * num for num in lista_numeros if num > 4]
print(resultado)
```

## Ejemplo integrador: preparar un resumen de pedidos (estilo realista)

Caso: una tienda tiene una lista de pedidos (diccionarios).
Queremos:
- filtrar pedidos pagados
- quedarnos con `total`
- calcular suma y media

Lo hacemos en versión “normal” y versión con herramientas funcionales.

**Datos de ejemplo:**

```python
pedidos = [
    {"id": 101, "cliente": "ana@example.com", "status": "paid", "total": 19.99},
    {"id": 102, "cliente": "luis@example.com", "status": "pending", "total": 55.00},
    {"id": 103, "cliente": "marta@example.com", "status": "paid", "total": 7.50},
    {"id": 104, "cliente": "ana@example.com", "status": "cancelled", "total": 12.00},
    {"id": 105, "cliente": "david@example.com", "status": "paid", "total": 120.00},
]
```

**Versión imperativa:**

```python
totales = []
for p in pedidos:
    if p["status"] == "paid":
        totales.append(p["total"])

suma = sum(totales)
if totales: # Lista con elementos True, si está vacía False
    media = suma / len(totales)
else:
    media = 0

print("Totales paid:", totales)
print("Suma:", suma)
print("Media:", media)
```

**Versión funcional (filter + map):**

```python
totales = list(map(lambda p: p["total"],
                   filter(lambda p: p["status"] == "paid", pedidos)))

suma = sum(totales)
media = suma / len(totales) if totales else 0

print("Totales paid:", totales)
print("Suma:", suma)
print("Media:", media)
```

**Versión con comprehension (habitualmente la más clara):**

```python
totales = [p["total"] for p in pedidos if p["status"] == "paid"]
print("Totales paid:", totales)
print("Suma:", sum(totales))
print("Media:", (sum(totales) / len(totales)) if totales else 0)
```

## Trampas comunes y buenas prácticas

- `map` y `filter` devuelven **iteradores**: si los imprimes, verás algo como `<map object ...>`.
  - Solución: `list(...)` (o consumirlos una vez).
- No “comprimir” por comprimir: a veces el bucle es más legible.
- Si la lambda empieza a ser larga, pásala a una función con `def`.
- Comprehensions:
  - suelen ser la mejor opción para listas
  - si la lógica crece demasiado, vuelve al bucle (o separa en funciones)
- Piensa en nombres: `nums`, `lista_numeros`, `pedidos`, etc. ayudan a leer.
