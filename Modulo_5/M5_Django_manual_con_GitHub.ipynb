## 1) Conceptos clave antes de empezar

### ¿Qué es Django?
**Django** es un framework de Python para crear aplicaciones web.

Te da muchas piezas ya resueltas:
- sistema de rutas
- plantillas HTML
- formularios
- autenticación de usuarios
- panel de administración
- ORM para trabajar con la base de datos

Dicho en lenguaje humano: Django te da el **esqueleto profesional** de una web para que no empieces desde cero.

### ¿Qué es un framework?
Un framework es una base de trabajo con reglas y herramientas ya preparadas.

En vez de tener una carpeta vacía y decidir tú toda la estructura, el framework te dice:
- dónde va cada cosa
- cómo se conectan las piezas
- qué comandos usar

### Arquitectura de Django: MVT (y su relación con MVC)
Muchos manuales hablan de **MVC**:
- **Model** → datos
- **View** → interfaz
- **Controller** → lógica que conecta todo

En Django normalmente hablamos de **MVT**:
- **Model** → define y representa los datos
- **View** → contiene la lógica de la petición/respuesta
- **Template** → define el HTML que verá el usuario

### Entonces, ¿dónde está el “Controller”?
En Django ese papel no aparece como una pieza única con ese nombre.

Se reparte entre:
- las **URLs**, que deciden qué vista se ejecuta
- las **views**, que procesan la petición
- el propio framework Django, que coordina todo

### Flujo mental de una petición en Django
1. el navegador pide una URL
2. Django busca esa URL en `urls.py`
3. llama a una vista en `views.py`
4. la vista consulta modelos o procesa formularios
5. la vista renderiza un template
6. el navegador recibe HTML

### ¿Qué es el ORM?
**ORM** significa **Object-Relational Mapping**.

Traducido a lenguaje humano:
> Te permite trabajar con la base de datos usando clases y objetos de Python en lugar de escribir SQL todo el rato.

##### Ejemplo mental

- Una clase `Trip` corresponde a una tabla  
- Un objeto `Trip(...)` corresponde a una fila  
- `Trip.objects.all()` ≈ “dame todos los registros”

##### Comparación SQL vs ORM

| SQL tradicional | ORM (Django) |
|----------------|-------------|
| `SELECT * FROM trips;` | `Trip.objects.all()` |
| `SELECT * FROM trips WHERE destination = 'París';` | `Trip.objects.filter(destination='París')` |
| `SELECT * FROM trips ORDER BY start_date DESC;` | `Trip.objects.order_by('-start_date')` |
| `SELECT * FROM trips WHERE destination = 'París' ORDER BY start_date DESC;` | `Trip.objects.filter(destination='París').order_by('-start_date')` |

La ventaja es que piensas en Python, no en sentencias SQL desde el minuto uno.

### ¿Qué es una migración?
Una **migración** es el mecanismo con el que Django actualiza la estructura de la base de datos.

Por ejemplo:
- añades un campo en un modelo
- Django genera una migración
- luego la aplicas para crear o modificar columnas/tablas

La idea corta es:
> **modelo cambiado en Python → migración → base de datos actualizada**

### ¿Qué es el panel de administración?
Django trae un **admin** ya hecho.

Sirve para:
- crear registros
- editar registros
- borrarlos
- gestionar usuarios

Es muy útil para enseñar, probar o administrar datos sin tener que construir primero una interfaz propia.

### ¿Qué es un CRUD?
**CRUD** son las 4 operaciones básicas sobre datos:
- **C**reate → crear
- **R**ead → leer o listar
- **U**pdate → editar
- **D**elete → borrar

### ¿Qué es un repositorio?
Un **repositorio Git** es la carpeta del proyecto con historial de versiones.

Git guarda:
- qué cambió
- cuándo cambió
- con qué mensaje
- a qué versión podemos volver

### ¿Qué papel tiene GitHub en esta práctica?
GitHub será la **copia remota y oficial** del proyecto.

La idea es ir subiendo cambios para que:
- quede guardado el progreso
- se practique GitHub de verdad
- podamos recuperar una versión estable si en local rompemos algo

## 6) Primera página real: dejar atrás la pantalla de bienvenida

Vamos a crear una página propia en vez de la página por defecto de Django.

### Qué estamos trabajando aquí
Este bloque corresponde sobre todo al **Tema 3**:
- rutas

### Flujo que vamos a montar
- `config/urls.py` delega en `trips/urls.py`
- `trips/views.py` prepara la respuesta
- un template HTML muestra la página

Es decir: **URL → vista → template**.

Esta es la parte más visible del patrón **MVT**.


### 6.1 Crear `trips/urls.py`

Este archivo contendrá las rutas propias de la app `trips`.

Aunque de momento solo tendremos la home, ya lo preparamos con `app_name` para poder usar nombres como `trips:home`.

````python
from django.urls import path   # Función para definir rutas
from . import views            # Importamos las vistas de esta misma app

app_name = 'trips'             # Espacio de nombres para usar rutas como trips:home

urlpatterns = [
    path('', views.home, name='home'),  # Ruta raíz de la app: llama a la vista home
]
````

### 6.2 Editar `config/urls.py`

Aquí conectamos las URLs generales del proyecto con las URLs internas de la app.

En lenguaje humano: le estamos diciendo a Django que, cuando alguien entre en `/`, mire las rutas definidas en `trips/urls.py`.

````python
from django.contrib import admin
from django.urls import path, include

urlpatterns = [
    path('admin/', admin.site.urls),  # Ruta del panel de administración
    path('', include('trips.urls')),  # Delegamos las rutas principales en la app trips
]
````

### 6.3 Editar `trips/views.py`

La vista `home` será una función sencilla que solo renderiza el template de inicio.

````python
from django.shortcuts import render  # Ayuda para devolver HTML renderizado con un template


def home(request):
    # Renderiza la plantilla de inicio y la devuelve como respuesta HTTP
    return render(request, 'trips/home.html')
````

### 6.4 Crear el template `trips/templates/trips/home.html`

Crea las carpetas si todavía no existen.

Este HTML será la primera página propia del proyecto. De momento será mínima, porque solo queremos comprobar que el flujo URL → vista → template funciona.

````python

<!doctype html>
<html lang="es">
<head>
  <meta charset="utf-8">
  <title>TravelChecklist</title>
</head>
<body>
  <h1>TravelChecklist</h1>
  <p>Tu gestor de viajes y tareas.</p>
</body>
</html>
````

### 8.2 Reemplazar `trips/views.py`

Ahora vamos a añadir la vista de registro.

De momento este archivo tendrá:
- `home` → página de inicio
- `signup` → formulario de registro y creación de usuario

````python
from django.contrib.auth.forms import UserCreationForm  # Formulario de registro ya preparado por Django
from django.shortcuts import render, redirect           # render devuelve HTML; redirect redirige a otra URL


def home(request):
    # Página de inicio pública
    return render(request, 'trips/home.html')


def signup(request):
    # Si el usuario envía el formulario, procesamos los datos
    if request.method == 'POST':
        form = UserCreationForm(request.POST)

        # Si el formulario es válido, guardamos el nuevo usuario
        if form.is_valid():
            form.save()
            return redirect('trips:login')  # Tras registrarse, lo enviamos al login
    else:
        # Si entra por GET, mostramos el formulario vacío
        form = UserCreationForm()

    # Renderizamos el template pasando el formulario al contexto
    return render(request, 'trips/signup.html', {'form': form})
````

### Idea importante
Cada viaje pertenece a un usuario.  
Por eso necesitaremos una relación con el modelo de usuario de Django.

### Piensa en el ORM así
Cuando definimos una clase `Trip(models.Model)`, no estamos creando solo una clase de Python:
también estamos describiendo **cómo será una tabla de la base de datos**.


### 9.1 Reemplazar `trips/models.py`

Aquí defines la estructura de la tabla `Trip`.

Piensa este archivo así:
- cada clase de modelo → una tabla
- cada atributo del modelo → una columna

````python
from django.conf import settings  # Para referenciar el modelo de usuario activo del proyecto
from django.db import models      # Base para definir modelos Django


class Trip(models.Model):
    # Cada viaje pertenece a un usuario
    user = models.ForeignKey(
        settings.AUTH_USER_MODEL,
        on_delete=models.CASCADE,   # Si se borra el usuario, se borran sus viajes
        related_name='trips'        # Permite acceder desde user.trips.all()
    )

    # Campos básicos del viaje
    title = models.CharField(max_length=120)        # Título del viaje
    destination = models.CharField(max_length=120)  # Destino
    start_date = models.DateField()                 # Fecha de inicio
    end_date = models.DateField()                   # Fecha de fin
    notes = models.TextField(blank=True)            # Notas opcionales
    created_at = models.DateTimeField(auto_now_add=True)  # Fecha de creación automática

    class Meta:
        # Orden por defecto al consultar viajes:
        # primero los que empiezan más tarde y, si empatan, los más recientes
        ordering = ['-start_date', '-created_at']

    def __str__(self):
        # Texto legible del objeto, útil en admin y depuración
        return f"{self.title} ({self.destination})"
````

### 10.1 Crear `trips/forms.py`

El formulario `TripForm` se apoyará en el modelo `Trip`.

### Idea importante
Con `ModelForm` Django puede generar un formulario a partir del modelo, ahorrándote mucho código repetitivo.

### Qué valida Django aquí
Aunque todavía no lo veamos con detalle, este formulario ya valida cosas como:
- tipos de campo
- campos obligatorios
- formato de fechas
- longitudes máximas


````python
from django import forms
from .models import Trip


class TripForm(forms.ModelForm):
    class Meta:
        model = Trip  # Modelo sobre el que se basa el formulario

        # Campos que queremos mostrar al usuario
        fields = ['title', 'destination', 'start_date', 'end_date', 'notes']

        # Widgets para mejorar la visualización del formulario
        widgets = {
            'start_date': forms.DateInput(attrs={'type': 'date'}),
            'end_date': forms.DateInput(attrs={'type': 'date'}),
            'notes': forms.Textarea(attrs={'rows': 4}),
        }
````

### 10.2 Reemplazar `trips/views.py`

Ahora el archivo `views.py` ya no solo tendrá la home y el registro.  
También tendrá las vistas del CRUD de viajes.

Fíjate especialmente en estas ideas:
- `@login_required` protege vistas para usuarios autenticados
- `get_object_or_404(...)` recupera un objeto o devuelve error 404 si no existe
- `request.method == 'POST'` permite distinguir envío de formulario frente a carga inicial

````python
from django.contrib.auth.decorators import login_required  # Obliga a iniciar sesión
from django.contrib.auth.forms import UserCreationForm
from django.shortcuts import render, redirect, get_object_or_404

from .forms import TripForm
from .models import Trip


def home(request):
    # Home pública
    return render(request, 'trips/home.html')


def signup(request):
    # Registro de usuario
    if request.method == 'POST':
        form = UserCreationForm(request.POST)
        if form.is_valid():
            form.save()
            return redirect('trips:login')
    else:
        form = UserCreationForm()

    return render(request, 'trips/signup.html', {'form': form})


@login_required
def trip_list(request):
    # Mostramos solo los viajes del usuario autenticado
    trips = Trip.objects.filter(user=request.user)
    return render(request, 'trips/trip_list.html', {'trips': trips})


@login_required
def trip_create(request):
    # Crear viaje
    if request.method == 'POST':
        form = TripForm(request.POST)
        if form.is_valid():
            trip = form.save(commit=False)  # Creamos el objeto sin guardar todavía
            trip.user = request.user        # Asignamos el usuario autenticado
            trip.save()
            return redirect('trips:trip_list')
    else:
        form = TripForm()

    return render(request, 'trips/trip_form.html', {'form': form, 'mode': 'create'})


@login_required
def trip_detail(request, pk):
    # Recupera el viaje del usuario o devuelve 404 si no existe/no le pertenece
    trip = get_object_or_404(Trip, pk=pk, user=request.user)
    return render(request, 'trips/trip_detail.html', {'trip': trip})


@login_required
def trip_update(request, pk):
    # Edición de un viaje existente
    trip = get_object_or_404(Trip, pk=pk, user=request.user)

    if request.method == 'POST':
        form = TripForm(request.POST, instance=trip)
        if form.is_valid():
            form.save()
            return redirect('trips:trip_detail', pk=trip.pk)
    else:
        form = TripForm(instance=trip)

    return render(request, 'trips/trip_form.html', {'form': form, 'mode': 'update', 'trip': trip})


@login_required
def trip_delete(request, pk):
    # Borrado de un viaje existente
    trip = get_object_or_404(Trip, pk=pk, user=request.user)

    if request.method == 'POST':
        trip.delete()
        return redirect('trips:trip_list')

    return render(request, 'trips/trip_confirm_delete.html', {'trip': trip})
````

## 11) Modelo `Task`: tareas dentro de un viaje

Ahora añadimos una segunda tabla relacionada con `Trip`.

### Idea importante
Cada tarea pertenece a un viaje.  
Eso significa que `Task` tendrá una `ForeignKey` hacia `Trip`.

### 11.1 Reemplazar `trips/models.py`

Ahora el archivo tendrá dos modelos: `Trip` y `Task`.

Observa la relación:
- un `Trip` puede tener muchas `Task`
- cada `Task` pertenece a un único `Trip`

````python
from django.conf import settings
from django.db import models


class Trip(models.Model):
    # Relación con el usuario propietario del viaje
    user = models.ForeignKey(
        settings.AUTH_USER_MODEL,
        on_delete=models.CASCADE,
        related_name='trips'
    )
    title = models.CharField(max_length=120)        # Título del viaje
    destination = models.CharField(max_length=120)  # Destino
    start_date = models.DateField()                 # Inicio del viaje
    end_date = models.DateField()                   # Fin del viaje
    notes = models.TextField(blank=True)            # Notas opcionales
    created_at = models.DateTimeField(auto_now_add=True)  # Fecha de creación

    class Meta:
        ordering = ['-start_date', '-created_at']

    def __str__(self):
        return f"{self.title} ({self.destination})"


class Task(models.Model):
    # Opciones cerradas para la prioridad
    class Priority(models.IntegerChoices):
        LOW = 1, 'Baja'
        MEDIUM = 2, 'Media'
        HIGH = 3, 'Alta'

    # Cada tarea pertenece a un viaje
    trip = models.ForeignKey(
        Trip,
        on_delete=models.CASCADE,  # Si se borra el viaje, se borran sus tareas
        related_name='tasks'       # Permite acceder desde trip.tasks.all()
    )

    title = models.CharField(max_length=160)  # Nombre de la tarea

    # Prioridad con valores controlados
    priority = models.IntegerField(
        choices=Priority.choices,
        default=Priority.MEDIUM
    )

    due_date = models.DateField(null=True, blank=True)  # Fecha límite opcional
    done = models.BooleanField(default=False)           # Indica si la tarea está hecha
    created_at = models.DateTimeField(auto_now_add=True)

    class Meta:
        # Primero las pendientes, luego las más prioritarias
        ordering = ['done', '-priority', 'due_date', '-created_at']

    def __str__(self):
        return self.title
````

### 12.1 Reemplazar `trips/forms.py`

Añadimos `TaskForm` manteniendo también `TripForm`.

````python
from django import forms
from .models import Trip, Task


class TripForm(forms.ModelForm):
    class Meta:
        model = Trip
        fields = ['title', 'destination', 'start_date', 'end_date', 'notes']
        widgets = {
            'start_date': forms.DateInput(attrs={'type': 'date'}),
            'end_date': forms.DateInput(attrs={'type': 'date'}),
            'notes': forms.Textarea(attrs={'rows': 4}),
        }


class TaskForm(forms.ModelForm):
    class Meta:
        model = Task
        fields = ['title', 'priority', 'due_date', 'done']
        widgets = {
            'due_date': forms.DateInput(attrs={'type': 'date'}),
        }
````

### 12.2 Reemplazar `trips/views.py`

Ahora `views.py` crecerá bastante porque incorporará también la lógica del CRUD de tareas.

No te agobies con el tamaño del archivo.  
La idea pedagógica aquí es que veas **toda la lógica junta** para entender bien el flujo.

````python
from django.contrib.auth.decorators import login_required
from django.contrib.auth.forms import UserCreationForm
from django.shortcuts import render, redirect, get_object_or_404

from .forms import TripForm, TaskForm
from .models import Trip, Task


def home(request):
    # Home pública
    return render(request, 'trips/home.html')


def signup(request):
    # Registro de usuario
    if request.method == 'POST':
        form = UserCreationForm(request.POST)
        if form.is_valid():
            form.save()
            return redirect('trips:login')
    else:
        form = UserCreationForm()

    return render(request, 'trips/signup.html', {'form': form})


@login_required
def trip_list(request):
    # Lista solo los viajes del usuario autenticado
    trips = Trip.objects.filter(user=request.user)
    return render(request, 'trips/trip_list.html', {'trips': trips})


@login_required
def trip_create(request):
    # Crear un nuevo viaje
    if request.method == 'POST':
        form = TripForm(request.POST)
        if form.is_valid():
            trip = form.save(commit=False)
            trip.user = request.user
            trip.save()
            return redirect('trips:trip_list')
    else:
        form = TripForm()

    return render(request, 'trips/trip_form.html', {'form': form, 'mode': 'create'})


@login_required
def trip_detail(request, pk):
    # Mostramos el viaje y todas sus tareas asociadas
    trip = get_object_or_404(Trip, pk=pk, user=request.user)
    tasks = trip.tasks.all()
    return render(request, 'trips/trip_detail.html', {'trip': trip, 'tasks': tasks})


@login_required
def trip_update(request, pk):
    # Editar un viaje existente
    trip = get_object_or_404(Trip, pk=pk, user=request.user)

    if request.method == 'POST':
        form = TripForm(request.POST, instance=trip)
        if form.is_valid():
            form.save()
            return redirect('trips:trip_detail', pk=trip.pk)
    else:
        form = TripForm(instance=trip)

    return render(request, 'trips/trip_form.html', {'form': form, 'mode': 'update', 'trip': trip})


@login_required
def trip_delete(request, pk):
    # Borrar un viaje
    trip = get_object_or_404(Trip, pk=pk, user=request.user)

    if request.method == 'POST':
        trip.delete()
        return redirect('trips:trip_list')

    return render(request, 'trips/trip_confirm_delete.html', {'trip': trip})


@login_required
def task_create(request, trip_pk):
    # Crear una tarea asociada a un viaje concreto del usuario
    trip = get_object_or_404(Trip, pk=trip_pk, user=request.user)

    if request.method == 'POST':
        form = TaskForm(request.POST)
        if form.is_valid():
            task = form.save(commit=False)
            task.trip = trip
            task.save()
            return redirect('trips:trip_detail', pk=trip.pk)
    else:
        form = TaskForm()

    return render(request, 'trips/task_form.html', {'form': form, 'trip': trip, 'mode': 'create'})


@login_required
def task_update(request, pk):
    # Editar una tarea; comprobamos que el usuario sea dueño del viaje al que pertenece
    task = get_object_or_404(Task, pk=pk, trip__user=request.user)

    if request.method == 'POST':
        form = TaskForm(request.POST, instance=task)
        if form.is_valid():
            form.save()
            return redirect('trips:trip_detail', pk=task.trip.pk)
    else:
        form = TaskForm(instance=task)

    return render(request, 'trips/task_form.html', {'form': form, 'trip': task.trip, 'task': task, 'mode': 'update'})


@login_required
def task_delete(request, pk):
    # Borrar tarea y volver al detalle del viaje
    task = get_object_or_404(Task, pk=pk, trip__user=request.user)
    trip = task.trip

    if request.method == 'POST':
        task.delete()
        return redirect('trips:trip_detail', pk=trip.pk)

    return render(request, 'trips/task_confirm_delete.html', {'task': task, 'trip': trip})
````

### 13.1 Reemplazar `trips/views.py`

Vamos a mejorar la vista `trip_list`.

### Qué cambia
- usaremos `Q` para buscar en varios campos
- usaremos `order_by()` para cambiar el orden
- usaremos `Paginator` para partir el listado en páginas

### Idea mental
Primero obtenemos una consulta base:

```python
Trip.objects.filter(user=request.user)
```

Y luego la vamos refinando:
1. filtramos
2. ordenamos
3. paginamos


````python
from django.contrib.auth.decorators import login_required
from django.contrib.auth.forms import UserCreationForm
from django.core.paginator import Paginator  # Divide resultados en páginas
from django.db.models import Q               # Permite búsquedas OR en varios campos
from django.shortcuts import render, redirect, get_object_or_404

from .forms import TripForm, TaskForm
from .models import Trip, Task


def home(request):
    # Home pública
    return render(request, 'trips/home.html')


def signup(request):
    # Registro de usuario
    if request.method == 'POST':
        form = UserCreationForm(request.POST)

        # form.is_valid() lanza la validación del formulario
        if form.is_valid():
            form.save()
            return redirect('trips:login')
    else:
        form = UserCreationForm()

    return render(request, 'trips/signup.html', {'form': form})


@login_required
def trip_list(request):
    # 1) Consulta base: solo viajes del usuario autenticado
    trips = Trip.objects.filter(user=request.user)

    # 2) Leemos parámetros GET del listado
    q = request.GET.get('q', '').strip()
    destination = request.GET.get('destination', '').strip()
    order = request.GET.get('order', 'start_desc')

    # 3) Búsqueda libre en varios campos
    if q:
        trips = trips.filter(
            Q(title__icontains=q) |
            Q(destination__icontains=q) |
            Q(notes__icontains=q)
        )

    # 4) Filtro por destino exacto (lo recibimos desde un desplegable)
    if destination:
        trips = trips.filter(destination=destination)

    # 5) Orden dinámico según la opción elegida
    order_options = {
        'start_desc': '-start_date',   # Más recientes primero
        'start_asc': 'start_date',     # Más antiguos primero
        'destination_asc': 'destination',
        'created_desc': '-created_at',
        'created_asc': 'created_at',
    }
    trips = trips.order_by(order_options.get(order, '-start_date'))

    # 6) Lista de destinos únicos para rellenar el filtro del template
    destinations = (
        Trip.objects
        .filter(user=request.user)
        .values_list('destination', flat=True)
        .distinct()
        .order_by('destination')
    )

    # 7) Paginación: 4 viajes por página
    paginator = Paginator(trips, 4)
    page_number = request.GET.get('page')
    page_obj = paginator.get_page(page_number)

    # 8) Enviamos al template tanto la página actual como los filtros activos
    context = {
        'trips': page_obj.object_list,
        'page_obj': page_obj,
        'q': q,
        'destination': destination,
        'order': order,
        'destinations': destinations,
    }
    return render(request, 'trips/trip_list.html', context)


@login_required
def trip_create(request):
    # Crear un nuevo viaje
    if request.method == 'POST':
        form = TripForm(request.POST)

        # Si el formulario pasa validación, guardamos el viaje
        if form.is_valid():
            trip = form.save(commit=False)
            trip.user = request.user
            trip.save()
            return redirect('trips:trip_list')
    else:
        form = TripForm()

    return render(request, 'trips/trip_form.html', {'form': form, 'mode': 'create'})


@login_required
def trip_detail(request, pk):
    # Mostramos el viaje y todas sus tareas asociadas
    trip = get_object_or_404(Trip, pk=pk, user=request.user)
    tasks = trip.tasks.all()
    return render(request, 'trips/trip_detail.html', {'trip': trip, 'tasks': tasks})


@login_required
def trip_update(request, pk):
    # Editar un viaje existente
    trip = get_object_or_404(Trip, pk=pk, user=request.user)

    if request.method == 'POST':
        form = TripForm(request.POST, instance=trip)
        if form.is_valid():
            form.save()
            return redirect('trips:trip_detail', pk=trip.pk)
    else:
        form = TripForm(instance=trip)

    return render(request, 'trips/trip_form.html', {'form': form, 'mode': 'update', 'trip': trip})


@login_required
def trip_delete(request, pk):
    # Borrar un viaje
    trip = get_object_or_404(Trip, pk=pk, user=request.user)

    if request.method == 'POST':
        trip.delete()
        return redirect('trips:trip_list')

    return render(request, 'trips/trip_confirm_delete.html', {'trip': trip})


@login_required
def task_create(request, trip_pk):
    # Crear una tarea asociada a un viaje concreto del usuario
    trip = get_object_or_404(Trip, pk=trip_pk, user=request.user)

    if request.method == 'POST':
        form = TaskForm(request.POST)
        if form.is_valid():
            task = form.save(commit=False)
            task.trip = trip
            task.save()
            return redirect('trips:trip_detail', pk=trip.pk)
    else:
        form = TaskForm()

    return render(request, 'trips/task_form.html', {'form': form, 'trip': trip, 'mode': 'create'})


@login_required
def task_update(request, pk):
    # Editar una tarea; comprobamos que el usuario sea dueño del viaje al que pertenece
    task = get_object_or_404(Task, pk=pk, trip__user=request.user)

    if request.method == 'POST':
        form = TaskForm(request.POST, instance=task)
        if form.is_valid():
            form.save()
            return redirect('trips:trip_detail', pk=task.trip.pk)
    else:
        form = TaskForm(instance=task)

    return render(request, 'trips/task_form.html', {'form': form, 'trip': task.trip, 'task': task, 'mode': 'update'})


@login_required
def task_delete(request, pk):
    # Borrar tarea y volver al detalle del viaje
    task = get_object_or_404(Task, pk=pk, trip__user=request.user)
    trip = task.trip

    if request.method == 'POST':
        task.delete()
        return redirect('trips:trip_detail', pk=trip.pk)

    return render(request, 'trips/task_confirm_delete.html', {'task': task, 'trip': trip})
````

### Idea importante
Los filtros los enviaremos por **GET**, no por POST.

¿Por qué?
Porque así la URL refleja el estado del listado, por ejemplo:

```text
/trips/?q=roma&order=start_desc&page=2
```

Eso facilita:
- compartir la URL
- recargar la página sin perder filtros
- entender mejor qué está pasando

## 17) Git y GitHub en acción: "la cagamos" y recuperamos el proyecto

Esta parte está pensada para hacerla **en directo en clase** porque suele ser muy útil.

La idea es enseñar dos escenarios distintos:

1. **He tocado un archivo y lo he roto, pero todavía no he hecho commit**  
2. **He dejado el proyecto hecho un desastre y quiero volver exactamente a la última versión oficial que está en GitHub**

> Importante: algunos comandos de esta sección descartan cambios. No los uses a ciegas en un proyecto real sin saber qué vas a perder.

### 17.1 Escenario 1: hemos roto un archivo local

Por ejemplo, imagina que editas `trips/views.py`, borras algo, guardas y de repente el proyecto deja de arrancar.

Lo primero es mirar qué ha cambiado.

````python
git status
````

Si el problema está en **un archivo concreto** y todavía no has hecho commit, puedes restaurarlo a la última versión guardada en tu repositorio local con:

````python
git restore trips/views.py
````

### Qué hace `git restore trips/views.py`
Descarta los cambios locales de ese archivo y lo deja exactamente como estaba en el último commit.

Esto viene muy bien cuando has roto **un fichero concreto** y quieres volver atrás sin tocar el resto del proyecto.

### 17.2 Escenario 2: queremos volver a la última versión oficial de GitHub

Ahora imagina algo peor:
- has tocado muchos archivos
- ya no sabes qué has roto
- quieres dejar el proyecto exactamente igual que la última versión subida a GitHub

En ese caso, la secuencia típica es esta:

````python
git fetch origin
git reset --hard origin/main
git clean -fd
````

### Qué hace cada comando
- `git fetch origin` → descarga la información más reciente del repositorio remoto, pero sin mezclarla todavía con tu copia local
- `git reset --hard origin/main` → coloca tu rama local exactamente en el mismo commit que `origin/main`
- `git clean -fd` → elimina archivos y carpetas no versionados que hayan quedado sueltos por el proyecto

### Resultado
Tu proyecto local queda como una copia de la última versión oficial subida a GitHub.

> Si tu rama principal se llama `master` en vez de `main`, sustituye `origin/main` por `origin/master`.

### 17.3 Cuándo usar cada estrategia

Usa **`git restore archivo`** cuando:
- el problema está en uno o pocos archivos
- todavía no has hecho commit
- no quieres tocar el resto del proyecto

Usa **`git fetch` + `git reset --hard` + `git clean -fd`** cuando:
- quieres una restauración total
- te da igual perder cambios locales no guardados
- quieres volver a la versión oficial de GitHub

### 17.4 Comandos útiles

Antes de restaurar, también puede ser útil estos dos comandos:

````python
git log --oneline --decorate --graph -n 10
git diff
````

### Qué enseñan estos comandos
- `git log --oneline --decorate --graph -n 10` → muestra de forma compacta los últimos commits y en qué rama estás
- `git diff` → enseña exactamente qué has cambiado respecto al último commit
