## Ejercicio 1: Eliminar duplicados con conjuntos

Dada la siguiente lista de nombres:

```python
nombres = ["Ana", "Luis", "Ana", "Marta", "Luis", "Carlos", "Elena"]
```

Crea un conjunto para eliminar los nombres repetidos.

Después, muestra:

1. La lista original.
2. El conjunto sin duplicados.
3. El número de nombres únicos.

````python
nombres = ["Ana", "Luis", "Ana", "Marta", "Luis", "Carlos", "Elena"]

nombres_unicos = set(nombres)

print("Lista original:", nombres)
print("Conjunto sin duplicados:", nombres_unicos)
print("Numero de nombres unicos:", len(nombres_unicos))
````

## Ejercicio 2: Operaciones entre conjuntos

Dos grupos de estudiantes se han apuntado a diferentes actividades:

```python
futbol = {"Ana", "Luis", "Carlos", "Marta"}
baloncesto = {"Luis", "Elena", "Marta", "Pedro"}
```

Calcula y muestra:

1. Los estudiantes que practican ambas actividades.
2. Los estudiantes que practican al menos una actividad.
3. Los estudiantes que practican fútbol pero no baloncesto.
4. Los estudiantes que practican solo una de las dos actividades.

````python
futbol = {"Ana", "Luis", "Carlos", "Marta"}
baloncesto = {"Luis", "Elena", "Marta", "Pedro"}

print("Ambas actividades:", futbol.intersection(baloncesto)) # baloncesto.intersection(futbol)
print("Al menos una actividad:", futbol.union(baloncesto))
print("Solo futbol:", futbol.difference(baloncesto))
print("Solo una actividad:", futbol.symmetric_difference(baloncesto))
````

## Ejercicio 5: Recorrer un diccionario con `.items()`

Dado el siguiente diccionario:

```python

    "pan": 1.20,
    "leche": 0.95,
    "huevos": 2.40,
```

Muestra cada producto junto con su precio usando el método `.items()`.

## Ejercicio 6: Uso de `.keys()`, `.values()` y `.get()`

Dado el siguiente diccionario:

```python

    "España": "Madrid",

    "Italia": "Roma",
```

Realiza las siguientes operaciones:

1. Muestra todos los países.

## Ejercicio 7: Análisis de frecuencia de palabras

Dado el siguiente texto:

```python
texto = """
Python es un lenguaje de programación muy utilizado.
Python permite trabajar con datos, automatizar tareas y crear aplicaciones.
Los datos son fundamentales en programación, análisis y ciencia de datos.
"""
```

Crea un programa que:

1. Convierta todo el texto a minúsculas.
2. Elimine puntos, comas y saltos de línea.
3. Separe el texto en palabras.
4. Cree un diccionario con la frecuencia de aparición de cada palabra.
5. Muestre solo las palabras que aparecen más de una vez.
6. Muestre la palabra más frecuente y su frecuencia.

````python
texto = """
Python es un lenguaje de programación muy utilizado.
Python permite trabajar con datos, automatizar tareas y crear aplicaciones.
Los datos son fundamentales en programación, análisis y ciencia de datos.
"""

texto = texto.lower()
texto = texto.replace(".", "")
texto = texto.replace(",", "")
texto = texto.replace("\n", " ")

palabras = texto.split() # split(" ") = split()

frecuencias = {}

for palabra in palabras:
    frecuencias[palabra] = frecuencias.get(palabra, 0) + 1

print("Palabras repetidas:")

for palabra, frecuencia in frecuencias.items():
    if frecuencia > 1:
        print(palabra, frecuencia)

palabra_mas_frecuente = ""
frecuencia_mas_frecuente = -999

# Esta es la forma habitual de calcular la palabra con más frecuencia:
for palabra, frecuencia in frecuencias.items():
    if frecuencia > frecuencia_mas_frecuente:
        palabra_mas_frecuente = palabra
        frecuencia_mas_frecuente = frecuencia

print(palabra_mas_frecuente, frecuencia_mas_frecuente)

# Esta es una versión equivalente, pero usando métodos:
palabra_mas_frecuente = max(frecuencias, key=frecuencias.get)
````

## Ejercicio 8: Clasificación de alumnos por curso y resultado

Dada la siguiente lista de diccionarios:

```python
alumnos = [
    {"nombre": "Ana", "nota": 8.5, "curso": "Python"},
    {"nombre": "Luis", "nota": 5.2, "curso": "Python"},
    {"nombre": "Marta", "nota": 9.1, "curso": "Data Science"},
    {"nombre": "Carlos", "nota": 4.8, "curso": "Python"},
    {"nombre": "Elena", "nota": 7.3, "curso": "Data Science"},
    {"nombre": "Pedro", "nota": 6.4, "curso": "Excel"}
]
```

Crea un programa que genere un diccionario con esta estructura:

```python
{
    "Python": {
        "aprobados": [...],
        "suspensos": [...]
    },
    "Data Science": {
        "aprobados": [...],
        "suspensos": [...]
    }
}
```

Un alumno estará aprobado si su nota es mayor o igual que `5`.

En las listas de aprobados y suspensos debe guardarse únicamente el nombre del alumno.

````python
alumnos = [
    {"nombre": "Ana", "nota": 8.5, "curso": "Python"},
    {"nombre": "Luis", "nota": 5.2, "curso": "Python"},
    {"nombre": "Marta", "nota": 9.1, "curso": "Data Science"},
    {"nombre": "Carlos", "nota": 4.8, "curso": "Python"},
    {"nombre": "Elena", "nota": 7.3, "curso": "Data Science"},
    {"nombre": "Pedro", "nota": 6.4, "curso": "Excel"}
]

clasificacion = {}

for alumno in alumnos:
    curso = alumno["curso"]

    if curso not in clasificacion:
        clasificacion[curso] = {
            "aprobados": [],
            "suspensos": []
        }

    if alumno["nota"] >= 5:
        clasificacion[curso]["aprobados"].append(alumno["nombre"])
    else:
        clasificacion[curso]["suspensos"].append(alumno["nombre"])

print(clasificacion)
````

## Ejercicio 9: Agrupación de correos electrónicos por dominio

Dada la siguiente lista de correos electrónicos:

```python
correos = [
    " ANA@GMAIL.COM ",
    "luis@yahoo.com",
    "marta@hotmail.com ",
    " carlos@gmail.com",
    "ELENA@GMAIL.COM",
    "pedro@yahoo.com ",
    "laura@empresa.com"
]
```

Crea un programa que:

1. Limpie cada correo eliminando espacios al principio y al final.
2. Convierta todos los correos a minúsculas.
3. Extraiga el dominio de cada correo.
4. Genere un diccionario donde:
   - La clave sea el dominio.
   - El valor sea una lista con todos los correos de ese dominio.

Ejemplo:

```python
{
    "gmail.com": ["ana@gmail.com", "carlos@gmail.com", "elena@gmail.com"],
    "yahoo.com": ["luis@yahoo.com", "pedro@yahoo.com"]
}
```

````python
correos = [
    " ANA@GMAIL.COM ",
    "luis@yahoo.com",
    "marta@hotmail.com ",
    " carlos@gmail.com",
    "ELENA@GMAIL.COM",
    "pedro@yahoo.com ",
    "laura@empresa.com"
]

correos_por_dominio = {}

for correo in correos:
    correo_limpio = correo.strip().lower()
    dominio = correo_limpio.split("@")[1]

    if dominio not in correos_por_dominio:
        correos_por_dominio[dominio] = []

    correos_por_dominio[dominio].append(correo_limpio)

print(correos_por_dominio)
````

## Ejercicio 10: Procesamiento de pedidos de una tienda

Dada la siguiente lista de pedidos:

```python
pedidos = [
    "Ana: portátil, ratón, teclado",
    "Luis: monitor, teclado",
    "Marta: portátil, monitor, ratón",
    "Carlos: ratón",
    "Elena: teclado, monitor"
]
```

Crea un programa que genere:

1. Un diccionario donde cada cliente sea una clave y el valor sea una lista con los productos comprados.
2. Un diccionario donde cada producto sea una clave y el valor sea el número total de veces que se ha comprado.
3. Una lista con los productos comprados más de una vez.

El programa debe limpiar correctamente los espacios sobrantes de los nombres de productos.

````python
pedidos = [
    "Ana: portátil, ratón, teclado",
    "Luis: monitor, teclado",
    "Marta: portátil, monitor, ratón",
    "Carlos: ratón",
    "Elena: teclado, monitor"
]

pedidos_clientes = {}
conteo_productos = {}

for pedido in pedidos:
    cliente, productos_texto = pedido.split(":")
    productos = productos_texto.strip().split(",")

    lista_productos = []

    for producto in productos:
        producto_limpio = producto.strip()
        lista_productos.append(producto_limpio)

        conteo_productos[producto_limpio] = conteo_productos.get(producto_limpio, 0) + 1

    pedidos_clientes[cliente] = lista_productos

productos_repetidos = []

for producto, cantidad in conteo_productos.items():
    if cantidad > 1:
        productos_repetidos.append(producto)

print("Pedidos por cliente:")
print(pedidos_clientes)

print("Conteo de productos:")
print(conteo_productos)

print("Productos repetidos:")
print(productos_repetidos)
````

## Ejercicio 11: Normalización y conteo de etiquetas

Dada la siguiente lista de etiquetas:

```python
etiquetas = [
    " Python ", "python", "PYTHON",
    "Data Science", " data science ",
    "Machine Learning", "machine learning",
    "IA", "ia", " Inteligencia Artificial "
]
```

Crea un programa que:

1. Elimine espacios al principio y al final de cada etiqueta.
2. Convierta todas las etiquetas a minúsculas.
3. Sustituya la etiqueta `"ia"` por `"inteligencia artificial"`.
4. Cree una lista final de etiquetas únicas.
5. Cree un diccionario donde:
   - La clave sea la etiqueta normalizada.
   - El valor sea el número de veces que aparecía originalmente.

La salida debe permitir saber cuántas veces aparece cada etiqueta después de normalizarla.

````python
etiquetas = [
    " Python ", "python", "PYTHON",
    "Data Science", " data science ",
    "Machine Learning", "machine learning",
    "IA", "ia", " Inteligencia Artificial "
]

conteo_etiquetas = {}

for etiqueta in etiquetas:
    etiqueta_limpia = etiqueta.strip().lower()

    if etiqueta_limpia == "ia":
        etiqueta_limpia = "inteligencia artificial"

    conteo_etiquetas[etiqueta_limpia] = conteo_etiquetas.get(etiqueta_limpia, 0) + 1

etiquetas_unicas = list(conteo_etiquetas.keys())

print("Etiquetas únicas:")
print(etiquetas_unicas)

print("Conteo de etiquetas:")
print(conteo_etiquetas)
````

## Ejercicio 12: Análisis de comentarios de clientes

Dada la siguiente lista de comentarios:

```python
comentarios = [
    "El producto es bueno, pero el envío fue lento",
    "Muy buen producto y excelente calidad",
    "El envío fue rápido y el producto llegó perfecto",
    "Mala calidad, no recomiendo este producto",
    "Buen precio y buena calidad"
]
```

Y el siguiente conjunto de palabras vacías:

```python
stopwords = {"el", "es", "y", "pero", "fue", "este", "no"}
```

Crea un programa que:

1. Convierta todos los comentarios a minúsculas.
2. Elimine las comas.
3. Separe cada comentario en palabras.
4. Ignore las palabras incluidas en `stopwords`.
5. Genere un diccionario con la frecuencia de aparición de cada palabra válida.
6. Ordene las palabras por frecuencia de mayor a menor.
7. Muestre las tres palabras más frecuentes.

````python
comentarios = [
    "El producto es bueno, pero el envío fue lento",
    "Muy buen producto y excelente calidad",
    "El envío fue rápido y el producto llegó perfecto",
    "Mala calidad, no recomiendo este producto",
    "Buen precio y buena calidad"
]

stopwords = {"el", "es", "y", "pero", "fue", "este", "no"}

frecuencias = {}

for comentario in comentarios:
    comentario_limpio = comentario.lower()
    comentario_limpio = comentario_limpio.replace(",", "")

    palabras = comentario_limpio.split()

    for palabra in palabras:
        if palabra not in stopwords:
            frecuencias[palabra] = frecuencias.get(palabra, 0) + 1

palabras_ordenadas = sorted(
    frecuencias.items(),
    key=lambda elemento: elemento[1],
    reverse=True
)

top_3 = palabras_ordenadas[:3]

print("Frecuencias:")
print(frecuencias)

print("Top 3 palabras más frecuentes:")

for palabra, frecuencia in top_3:
    print(palabra, frecuencia)
````
